```python
def simple_interest(p,r,t):
    simple_intrest=(p*r*t)/100
    return simple_interest
```

```python
def compound_intrest(p,r,t):
    amount=p=(1+(r/100))**t
    compound_intrest= amount - p
    return compound_intrest
```

```python
n=int(input("enter the number of customers"))
for i in range(n):
    print("\n customers",i+1)
    p=int(input("enter the principle"))
    r=int(input("enter the rate"))
    t=int(input("enter the time"))
    simple_interest=simple_interest(p,r,t)
    compound_intrest=compound_intrest(p,r,t)
    print("simple_interest",simple_interest)
    print("compound_intrest",compound_intrest)
```
