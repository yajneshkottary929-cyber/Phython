```python
import pandas as pd 
from openpyxl import Workbook
```

```python
df=pd.read_csv("transactions.csv")
```

```python
df
```

```python
df["fraud_Status"]="Normal"
```

```python
df.loc[
    (df["Amount"]>50000)|
    (df["Amount"]<0)|
    (df["Transaction_Count"]>5),
    "Fraud_Status"
]="Suspicion"     
```

```python
df
```

```python
wb=Workbook()
ws=wb.active
ws.title="Fraud Report"
```

```python
for col_num,column_name in enumerate(df.columns, 1):
    ws.cell(row=1,column=col_num,value=column_name)
```

```python
for raw_num , row in enumerate(df.values,2):
    for col_num,value in enumerate(row,1):
        ws.cell(row=raw_num,column=col_num,value=value)
```

```python
wb.save("Fraud_Detection_report.xlse")
print("excel report generated successfully")
```
