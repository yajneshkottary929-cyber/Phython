```python
df["Current_ratio"]=df["Current_Assets"]/df["Current_Liabilities"]
```

```python
df["RDE"]=(df["Net_Income"]/df["Equity"]*100)
```

```python
df["EPS"]=(df["Net_Income"]/df["Shares_Outstanding"]*100)
```

```python
df["EPS"]
```

```python
min=df["Current_ratio"].min()
print("minimum ratio",min)
```

```python
min=df["RDE"].min()
print("RDE",min)
```

```python
max=df["EPS"].min()
print("EPS",min)
```

```python
plt.figure(figsize=(12,7))
plt.plot(df["Year"],df["EPS"],marker="o")
plt.title("EPS tend")
plt.ylabel("Year")
plt.xlabel("EPS")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,7))
plt.plot(df["Year"],df["EPS"],marker="o")
plt.title("RDS tend")
plt.ylabel("Year")
plt.xlabel("RDS")
plt.grid(True)
plt.show()
```
