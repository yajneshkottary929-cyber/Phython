```python
import pandas as pd
```

```python
df=pd.read_csv("financial_dataset.csv")
```

```python
df
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["Profit"]=df["Profit"].fillna(df["Profit"].mean())
```

```python
df["Sales"]=df["Sales"].fillna(df["Sales"].mean())
```

```python
df["Payment_Status"]=df["Payment_Status"].fillna(df["Payment_Status"].mode()[0])
```

```python
df["City"]=df["City"].fillna(df["City"].mode()[0])
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df["City"]=df["City"].str.strip().str.title()	
```

```python
df["Category"]=df["Category"].str.strip().str.title()	
```

```python
df["Payment_Status"]=df["Payment_Status"].str.strip().str.title()	
```

```python
df["Quantity"]=pd.to_datetime(df["Quantity"],errors="coerce")
```
